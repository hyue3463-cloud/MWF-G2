## 3. Overall Recommendations for the Client

### Recommendation 1: Prioritise simple and actionable public-facing indicators

The client should prioritise indicators that the public can understand and act on quickly:

- **temperature** for general thermal comfort;
- **humidity** for uncomfortable or oppressive weather messages;
- **precipitation** for rainfall alerts;
- **wind speed** for general wind conditions;
- **gust** for sudden wind risk warnings;
- **UV index** for sun exposure warnings.

These indicators are easier to translate into public guidance.

### Recommendation 2: Use technical indicators as supporting context

Some indicators are useful but less intuitive for the public:

- **wind chill** may add limited value in Malaysia's tropical climate;
- **dew point** can support technical interpretation of muggy conditions;
- **pressure** may provide internal context for rainfall conditions but should not be used alone as a public rainfall warning.

### Recommendation 3: Communicate UV risk using categories

The client should communicate UV index using standard categories:

| UV Index | Category |
|---:|---|
| 0–2 | Low |
| 3–5 | Moderate |
| 6–7 | High |
| 8–10 | Very High |
| 11+ | Extreme |

These categories can be paired with action advice, such as seeking shade, using sunscreen and reducing direct sun exposure during high-risk periods.

## 4. Data Management Recommendations

To improve future public weather communication, the client should improve the underlying data management process.

Recommended improvements:

1. **Expand geographic coverage**  
   Collect data from more Malaysian states and locations so findings can better represent Malaysia as a whole.

2. **Reduce missing values**  
   Improve collection consistency for variables such as UV index, gust and precipitation.

3. **Preserve imputation flags**  
   If missing values are filled, keep columns that show whether each value was observed or imputed.

4. **Standardise units and definitions**  
   Ensure that variables such as wind speed, gust and precipitation have clear and consistent units.

5. **Add air quality variables**  
   The data dictionary mentions air quality, but the available CSV does not include air quality fields. Adding air quality would support more complete public health-related communication.

## 6. Limitations

The conclusions should be interpreted carefully because of several limitations:

1. **Limited geographic coverage**  
   The data covers selected locations in Kuala Lumpur and Pulau Pinang only. The findings should not be generalised to all of Malaysia.

2. **Missing values in the raw dataset**  
   Several key variables had missing values before cleaning. If values were imputed, some patterns may partly reflect the cleaning method.

3. **Uneven location representation**  
   Some cities or places have many more records than others, which may affect comparisons.

4. **Data dictionary mismatch**  
   The data dictionary mentions air quality, but the CSV does not include air quality variables. This limits the project's ability to assess air-quality-related public communication.

5. **Associations, not causation**  
   Relationships between variables should be interpreted as associations. For example, pressure may be associated with rainfall conditions, but the analysis does not prove direct causation.

6. **Simplified public communication framework**  
   The project evaluates usefulness based on data patterns and interpretability, but it does not test actual public understanding through surveys or experiments.
